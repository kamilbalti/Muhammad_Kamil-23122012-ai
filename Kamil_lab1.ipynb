```python
cities = [
    ["Quetta", 4000, 8000, 19000],
    ["Gwadar", 4500, 8500, 18000],
    ["Lahore", 5000, 7000, 17000],
    ["Islamabad", 6500, 9000, 20000],
    ["Murree", 7500, 12000, 25000],
    ["Muzaffarabad", 8500, 12500, 26000],
    ["Peshawar", 6000, 9000, 20000],
    ["Swat", 8500, 12000, 25000],
    ["Hunza", 11000, 15000, 30000],
    ["Skardu", 13000, 17000, 35000]
]

packages = [
    ["BASIC", "🚆 Train", 3000, 1500, "3-Star"],
    ["STANDARD", "🚌 Bus", 6000, 2500, "4-Star"],
    ["PREMIUM", "✈️ Plane", 12000, 4000, "5-Star"]
]

print("\n==========================================")
print("          KARACHI TRAVEL AGENT")
print("==========================================")

print("\nAvailable destinations:\n")

for city, basic_ticket, standard_ticket, premium_ticket in cities:
    print(f"- {city}")

destination = input(
    "\nWhere do you want to travel? "
).lower().strip()

selected_city = None

for city, basic_ticket, standard_ticket, premium_ticket in cities:
    if city.lower() == destination:
        selected_city = city
        break

if selected_city is None:
    print("\n❌ City not found.")

else:
    budget = float(input("Enter your budget (PKR): "))
    days = int(input("Enter number of days: "))
    people = int(input("Enter number of people: "))

    nights = max(days - 1, 1)
    results = []

    # Get ticket prices for selected city
    if selected_city == city:
        pass

    for city, basic_ticket, standard_ticket, premium_ticket in cities:
        if city == selected_city:
            ticket_prices = [
                basic_ticket,
                standard_ticket,
                premium_ticket
            ]
            break

    for i, package in enumerate(packages):

        name = package[0]
        transport = package[1]
        ticket_price = ticket_prices[i]
        hotel_price = package[2]
        food_price = package[3]
        stars = package[4]

        ticket = ticket_price
        hotel = hotel_price * nights
        food = food_price * days

        total_person = ticket + hotel + food
        total = total_person * people

        results.append([name, total])

    print("\n==========================================")
    print("               YOUR TRIP")
    print("==========================================")

    print(f"Destination: {selected_city}")
    print(f"Days: {days}")
    print(f"People: {people}")
    print(f"Budget: PKR {budget:,.0f}")

    for i, package in enumerate(packages):

        name = package[0]
        transport = package[1]
        stars = package[4]

        ticket_price = ticket_prices[i]
        hotel_price = package[2]
        food_price = package[3]

        ticket = ticket_price
        hotel = hotel_price * nights
        food = food_price * days

        total = (ticket + hotel + food) * people

        print("\n------------------------------------------")
        print(name)
        print(f"{transport} + {stars} Hotel")
        print("------------------------------------------")

        print(f"Ticket: PKR {ticket * people:,.0f}")
        print(f"Hotel: PKR {hotel * people:,.0f}")
        print(f"Food: PKR {food * people:,.0f}")
        print(f"TOTAL: PKR {total:,.0f}")

        if total <= budget:
            print("Status: ✅ Affordable")
        else:
            print("Status: ❌ Over Budget")

    basic_total = results[0][1]
    standard_total = results[1][1]
    premium_total = results[2][1]

    print("\n==========================================")
    print("              RECOMMENDATION")
    print("==========================================")

    if premium_total <= budget:

        print("⭐ Recommendation: Go with PREMIUM!")
        print(
            f"You have PKR "
            f"{budget - premium_total:,.0f} remaining."
        )

    elif standard_total <= budget:

        extra = premium_total - budget

        print("⭐ Recommendation: Go with STANDARD.")
        print(
            f"If you add PKR {extra:,.0f}, "
            "you can enjoy PREMIUM."
        )

    elif basic_total <= budget:

        extra = standard_total - budget

        print("⭐ Recommendation: Go with BASIC.")
        print(
            f"If you add PKR {extra:,.0f}, "
            "you can upgrade to STANDARD."
        )

    else:

        extra = basic_total - budget

        print("❌ None of the packages fit your budget.")
        print(
            f"You need PKR {extra:,.0f} more "
            "for BASIC."
        )

print("\n==========================================")
print("             END OF PROGRAM")
print("==========================================")
```
